# Dataset — Predictive Maintenance Aircraft Engine

## Objectif
Ce dataset est utilisé pour faire de la **maintenance prédictive** sur des moteurs d’avion.
L’objectif est de prédire le **nombre de cycles restants avant la panne** (*Remaining Useful Life*).

---

## Structure des données

Le dataset est composé de **trois fichiers** :

### 1. PM_train.csv
- Données d’entraînement
- Chaque moteur est observé cycle après cycle
- Un cycle correspond à une phase de fonctionnement
- À chaque cycle, 21 capteurs sont mesurés
- Les moteurs commencent avec une usure initiale différente (inconnue)
- La dernière ligne de chaque moteur correspond à la panne

---

### 2. PM_test.csv
- Même structure que les données d’entraînement
- Les moteurs ne vont pas jusqu’à la panne
- Le nombre de cycles restants est inconnu

---

### 3. PM_truth.csv
- Contient la vraie valeur du nombre de cycles restants
- Sert à évaluer les performances du modèle

---

## Nature du problème
- Problème de **séries temporelles multivariées**
- Chaque moteur correspond à une séquence temporelle
- La variable cible est le **RUL** (Remaining Useful Life)

---

## Capteurs
- Certains capteurs sont constants et peuvent être supprimés
- D’autres varient légèrement autour de 0
- Les capteurs physiques (température, pression, vitesse) sont les plus informatifs

---

## Enjeu data science
